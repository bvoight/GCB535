### 2f. Pull a change made on GitHub

So far every change has started on CoCalc and been *pushed* up. Changes can also start on the remote — a collaborator pushes something, or you edit a file on github.com directly — and then you need to *pull* them down. Let's do the second one, and run into a very common complication along the way.

**Make a change on GitHub.**

1. On your repo's page on github.com, click `README.md`, then the pencil icon (**Edit this file**).
2. Add a line **directly under the title** (the first line), e.g. `Edited on GitHub to test git pull.`
3. Click **Commit changes…**, give it a message such as `Edit README on GitHub`, and click **Commit changes**.

GitHub has just made a commit that your CoCalc clone does not have.

**Meanwhile, back on CoCalc…** suppose you were in the middle of some work and had edited the README locally too, but hadn't committed it. Simulate that. In the terminal, execute:

```
$> echo "Work in progress - not ready to commit" >> README.md
$> git status
```

`git status` reports `README.md` as modified, and also says `Your branch is up to date with 'origin/main'`. The second part is wrong — but git is not lying, it just hasn't looked. `git status` compares against the copy of the remote it saw *last time it talked to GitHub*. Ask it to check again:

```
$> git fetch
$> git status
```

Now it reports `Your branch is behind 'origin/main' by 1 commit, and can be fast-forwarded`. `git fetch` downloads what the remote has without touching your files; `git status` then tells you where your local copy stands — ahead (commits to push), behind (commits to pull), or both.

**Try to pull.** In the terminal, execute:

```
$> git pull
```

It refuses: `error: Your local changes to the following files would be overwritten by merge: README.md`. git will not merge incoming commits into a file you have uncommitted edits in, because it has no way to protect those edits. You will hit this constantly in real work: a half-finished change sitting in your working directory, and you need a clean pull.

**Stash, pull, unstash.** `git stash` takes every uncommitted change, puts it on a shelf, and leaves the working directory clean — exactly as of the last commit. In the terminal, execute:

```
$> git stash
$> git status
$> cat README.md
```

The working directory is clean and your scratch line is gone from the README (it's on the shelf, not deleted). Now the pull works:

```
$> git pull
$> git log --oneline
$> cat README.md
```

6 commits (up from 5), and the GitHub line is in your local README. Finally, take your work back off the shelf:

```
$> git stash pop
$> git status
$> cat README.md
```

Both edits are present: the GitHub line near the top, your scratch line at the bottom, and `git status` shows `README.md` modified again — right where you were before, except now you're up to date with the remote.

That scratch line was only a stand-in for real work, so discard it. In the terminal, execute:

```
$> git restore README.md
$> git status
```

**Rules of thumb:** `git pull` before you start working, `git push` when you stop. If you have uncommitted changes and need to pull, `git stash` → `git pull` → `git stash pop`. (`git stash list` shows what's on the shelf if you forget.)

## Part 3 — Time travel: recovering from a mistake

We are going to break the script, commit the broken version (as happens in real life), and then get the working version back.

### 3a. Break it

Replace the whole script with one line, confirm it is broken, and commit it. In the terminal, execute:

```
$> echo "print('oops, I deleted everything')" > gc_content.py
$> python gc_content.py example.fasta
$> git add gc_content.py
$> git commit -m "Refactor GC script"
```

Note the misleadingly cheerful commit message — also realistic.

### 3b. Find the last good version

In the terminal, execute:

```
$> git log --oneline
```

The top commit is the broken one. The one below it is the GitHub README edit, and below that is your last Part 2e version of the script, which worked. Note its short hash (the 7 characters at the left, e.g. `a1b2c3d`).

See exactly what the bad commit did. In the terminal, execute:

```
$> git diff HEAD~1 HEAD
```

`HEAD` means "the latest commit"; `HEAD~1` is the one before it. The diff shows everything you destroyed.

### 3c. Restore the file from the good commit

In the terminal, execute (substituting the hash you noted):

```
$> git checkout <good-hash> -- gc_content.py
$> python gc_content.py example.fasta
$> git status
```

The file is back, and `git status` shows it as a *staged* change (because you pulled it from history, git has already staged it for you). Commit the recovery and push. In the terminal, execute:

```
$> git commit -m "Restore working GC script after bad refactor"
$> git push
```

You did *not* delete the bad commit. History is append-only — the mistake is still recorded, and so is the fix. That is the point: nothing is ever lost.

````python
# Verification: 8 commits; the script runs again
!cd ~/biom5350/biom5350-seq-tools && git log --oneline && echo "---" && python gc_content.py example.fasta | head -n 3
````

## Bonus — Part 6: try out collaborating with a classmate

Congrats on making it this far! At this point, you're geared up to write, preserve, and manage your own code repository. In many cases, this will be entirely your own (and on your own). However, for a development or collaborative project, you can have multiple people creating, editing, pushing code onto a single code base. This is where git really shines! If you are up for it, find a colleague nearby and ask to share code and play out a couple of scenarios with them:

**Scenario 1 — a collaborator pushes to your repo.**

1. **Person A** adds **Person B** as a collaborator: on A's repo, *Settings ▸ Collaborators ▸ Add people*. B accepts the invitation (check email or github.com notifications).
2. **B clones A's repo** into `~/biom5350/`. It will land in a folder with the same name as B's own repo, so clone with a different target name: `git clone <url> partner-repo`.
3. **B makes a change** — add their name to the README under a *Contributors* heading — commits, and pushes. Because B is a collaborator, the push succeeds.
4. **A runs `git pull`** and sees B's commit appear in `git log`.
5. Swap roles.

**Scenario 2 — a merge conflict.** Both A and B edit the *same line* of the README (e.g. the title) in their own clones. A commits and pushes first. B commits, then tries to push — it is **rejected** because the remote has a commit B doesn't have. B runs `git pull` and git reports a **merge conflict**. Open the README: git has marked the conflicting lines with `<<<<<<<`, `=======`, `>>>>>>>`. Edit the file to keep whichever version you want (delete the markers), then `git add README.md`, `git commit`, `git push`.

You have now done the single most feared thing in git, and survived.
